### ==> 3. Data Transformation
***

**Create occupancy percentage column**

```python
df_agg_bookings.head(3)
```

```python
df_agg_bookings['occ_pct'] = df_agg_bookings.apply(lambda row: row['successful_bookings']/row['capacity'], axis=1)
```

You can use following approach to get rid of SettingWithCopyWarning

```python
new_col = df_agg_bookings.apply(lambda row: row['successful_bookings']/row['capacity'], axis=1)
df_agg_bookings = df_agg_bookings.assign(occ_pct=new_col.values)
df_agg_bookings.head(3)
```

Convert it to a percentage value

```python
df_agg_bookings['occ_pct'] = df_agg_bookings['occ_pct'].apply(lambda x: round(x*100, 2))
df_agg_bookings.head(3)
```

### ==> 5. Key Findings
***

- **Occupancy by room class:** Presidential (RT4) leads at 59.28%, slightly ahead of Premium (58.03%), Elite (58.01%) and Standard (57.88%) — fairly even across classes.
- **Occupancy by city:** Delhi is highest at 61.51%, followed by Hyderabad (58.12%), Mumbai (57.90%) and Bangalore (56.33%).
- **Weekday vs weekend occupancy:** Weekends run much higher — 72.34% vs 50.88% on weekdays, a ~21-point gap.
- **Revenue by city:** Mumbai leads by a wide margin (₹668.6M), then Bangalore (₹420.4M) and Hyderabad (₹325.2M), with Delhi lowest (₹294.4M) despite having the best occupancy and rating.
- **Revenue by month:** May 22 was strongest (₹581.8M), Jul 22 close behind (₹572.8M), Jun 22 lowest (₹553.9M).
- **Revenue by property:** Atliq Exotica is the top performer (₹320.3M); Atliq Seasons trails badly (₹66.1M) — under a quarter of the next-lowest property.
- **Average rating by city:** Delhi highest (3.78), then Hyderabad (3.66), Mumbai (3.65), Bangalore lowest (3.41).
- **Revenue by booking platform:** "Others" dominates (₹699.4M), then MakeYourTrip (₹340.8M), Logtrip (₹187.6M), Direct Online (₹169.0M), Tripster (₹123.1M), Journey (₹102.5M), Direct Offline lowest (₹86.4M).

**Takeaway:** Delhi wins on occupancy and guest satisfaction but underperforms on revenue, while Mumbai shows the opposite pattern — suggesting a pricing/positioning gap worth investigating. Direct booking channels (online + offline combined) bring in far less revenue than third-party platforms, which is worth flagging given the acquisition-cost implications.
